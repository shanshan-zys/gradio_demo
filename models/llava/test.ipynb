```python
pretrained_weight = '/home/zhenghuang/forgery_detection/LLaVA/pretrained_weights/raw_test_epoch5_lr0001/llava_clip_contrast_2025-10-06-17-28/epoch_ckpt/epoch1.pth'
pretrained_weight = torch.load(pretrained_weight)
```

```python
new_state_dict = OrderedDict()
for key, value in pretrained_weight.items():
    # 筛选出键名包含 "model.vision_model." 的条目
    if "model.vision_model." in key:
        # 将 "model.vision_model." 替换为 "vision_model."
        new_key = key.replace("model.vision_model.", "vision_model.")
        new_state_dict[new_key] = value
```
